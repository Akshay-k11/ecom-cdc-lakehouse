```python
import os, sys
import pytest

sys.dont_write_bytecode = True   # don't create __pycache__ folders inside the Git folder
```

```python
exit_code = pytest.main([os.path.abspath("../tests"), "-v", "-p", "no:cacheprovider"])
assert exit_code == 0, "Some tests failed. See the output above."

#   - -v prints one line per test.
#   - -p no:cacheprovider stops pytest from creating a .pytest_cache folder.
#   - The last assert turns the cell red when any test fails.
```
